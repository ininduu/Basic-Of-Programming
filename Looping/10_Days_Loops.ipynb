# Loops

1. while loop

# While Loop

- We use the reserved word while to make a while loop. It is used to execute a block of statements repeatedly until a given condition is satisfied. When the condition becomes false, the lines of code after the loop will be continued to be executed.

```python
# Example 1

count = 10
while count < 15:
  print(count)
  count = count + 1
```

```python
# Example 2

number = 50
while number < 60:
  print(number)
  number = number + 1
```

In the above while loop, the condition becomes false when count is 5. That is when the loop stops. If we are interested to run block of code once the condition is no longer true, we can use else.

```python
rr = 4
while rr < 10:
  print(rr)
  rr = rr +1
```

```python
nbr = 10
while nbr < 20:
  print(nbr)
  nbr = nbr + 1
else:
  print(nbr)
```

```python
# Example 2 - When variable testinng as samme in < 57

testing = 57
while testing < 57:
  print(testing)
  testing = testing + 1
else:
  print("")
```

The above loop condition will be false when count is 5 and the loop stops, and execution starts the else statement. As a result 5 will be printed.

### Break and Continue - Part 1
- Break: We use break when we like to get out of or stop the loop.

```python
# Example 1 : Using Break

nmb = 25
while nmb < 32:
  print(nmb)
  nmb = nmb + 1
  if nmb == 30:
    break
```

```python
# Example 2 : Using Break

united = 82
while united < 90:
  print(united)
  united = united + 1
  if united == 86:
    break
```

```python
mosue = 42
while mosue < 50:
  print(mosue)
  mosue = mosue + 1
  if mosue == 45:
    break
```

```python
# Example 3 : Using Break

city = 18
while city < 30:
  print(city)
  city = city + 1
  if city == 26:
    break
```

```python
# Example 1 : using Continue

hgh = 21
while hgh < 28:
  if hgh == 25:
    hgh = hgh + 1
    continue
  print(hgh)
  hgh = hgh + 1
```

```python
# Example 2 : using Continue

joss = 6
while joss < 15 :
  if joss == 10:
    joss = joss + 1
    continue
  print(joss)
  joss = joss + 1
```

```python
# Example 3 : Using Continue

sans = 5
while sans < 12:
  if sans == 9:
    sans = sans + 1
    continue
  print(sans)
  sans = sans + 1
```

The above while loop only prints 0, 1, 2 and 4 (skips 3).

# For Loop

- A for keyword is used to make a for loop, similar with other programming languages, but with some syntax differences. Loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

- Using For loop on list

```python
# Example 1

realMadrid = ['Rudiger', 'Arda Guler', 'Bellingham', 'Alvaro Careras']
for team in realMadrid:
  print(team)
```

```python
# Example 2

angka = [8, 6, 9, 10, 11]
for spesial in angka:
  print(spesial)
```

### Using For loop on string

```python
country = ['Spain', 'Germany', 'Italy', 'France']
for test in country:
  print(test)
```

```python
# Example 2

country = 'Argentina'
for i in range(len(country)):
  print(country[i])
```

```python
# Segitiga A

afc = "A"
for i in range(1, 6):
  print("A" * i)
```

```python
cara = '*'
for i in range(1, 8):
  print(cara * i)
```

### Using For loop on tuple

```python
num = (0, 1, 2, 3, 4, 5)
for number in num:
  print(number)
```

## Break and Continue - Part 2

- Short reminder: Break: We use break when we want to stop our loop before it is completed.

```python
# Break 1

trans = (0, 1, 2, 3, 4, 5, 6, 7)
for item in trans:
    print(item)
    if item == 4:
      break
```

```python
# break 2

asolole = (10, 11, 12, 13, 14, 15, 16, 17)
for lole in asolole:
  print(lole)
  if lole == 13:
    break
```

```python
# Continue 1

jepit = (44, 45, 46, 47, 48, 49, 50)
for jos in jepit:
  if jos == 47:
    jos = jos + 1
    continue
  print(jos)
  jos = jos + 1
```

```python
smua = (33, 34, 35, 36, 37, 38, 39, 40)
for cok in smua:
  if cok == 36:
    cok = cok + 1
    continue
  print(cok)
  cok = cok + 1
```

```python
numbers = (0,1,2,3,4,5)
for number in numbers:
    print(number)
    if number == 3:
        continue
    print('Next number should be ', number + 1) if number != 5 else print("loop's end") # for short hand conditions need both if and else statements
print('outside the loop')
```

In the example above, if the number equals 3, the step after the condition (but inside the loop) is skipped and the execution of the loop continues if there are any iterations left.

# Nested For Loop

```python
person = {
    'first_name': 'Chester',
    'last_name': 'Benington',
    'age': 24,
    'country': 'California',
    'is_marred': True,
    'skills': ['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}

def simpan(person):
  for key in person:
    if key == 'skills':
      for skill in person['skills']:
        print(skill)

simpan(person)
```

# For Else

- If we want to execute some message when the loop ends, we use else.

```python
for jos in range(8):
  print(jos)
else:
  print("This range goona stop at : ", jos)
```

# Pass

```python
for number in range(6):
    pass
```

```python
# Soal 1 — Absen Kelas
# Cerita:
# Kamu adalah ketua kelas. Ada 10 siswa yang harus dicek kehadirannya satu per satu

smager = 1
while smager < 11:
  print(f"Siswa ke {smager} Hadir ")
  smager = smager + 1
else:
  print("Semua Siswa telah absen")
```

```python
# Soal 2 — Cek Struk Belanja
# Cerita:
# Kamu belanja di minimarket. Daftar harga barang:

cok = [12500, 5000, 20500, 50000, 150000]
hasil = 0
for we in cok:
  hasil += hasil + we
  hasil += 0

print(f"Hasil Dari Keseluruhan {hasil}")
```
