## Break and Continue

```python
count = 0
while count < 10:
    print(count)
    count = count + 1
    if count == 8:
        break
```

```python
you = 0
while you < 20 :
    print(you)
    you = you + 1
    if you == 10:
        break
```

```python
# While looping using break and using continue

malang = 0
while malang < 15:
    if malang == 10:
        malang += 1
        continue
    print(malang)
    malang = malang + 1
```

## The question

### The user has the chance to enter the PIN 3 times. If they keep getting it wrong, the program immediately breaks with the account blocked status.

### After the PIN is correct, enter the transaction menu (nested loop or main transaction loop).

```python
pin = 8989
pinAttempt = 0
pinMaxAttempt = 3

# Looping
while pinAttempt < pinMaxAttempt:

    # user Input
    user = int(input("Enter your PIN : "))


    # Conditional
    if user == pin:
      print("You can continue on the Menu")
      break

    else:
      pinAttempt += 1

      if pinAttempt == pinMaxAttempt:
        print("Your account is blocked")
        break
      else:
        print("try again")
```

# Sistem Validasi & Agregasi Data Stream Ber-Sentinel (Medium)

Buatlah sebuah program yang mensimulasikan pembacaan stream data angka masuk secara terus-menerus menggunakan while loop.

- Program harus mengabaikan angka negatif dengan menggunakan continue.

- Program harus berhenti total (terminate) jika mendapati angka 999 menggunakan break.

- Jika angka valid (positif dan bukan 999), masukkan ke dalam perhitungan jumlah kumulatif. Namun, jika angka tersebut adalah kelipatan 5, abaikan proses penambahan ke total kumulatif (gunakan continue atau logika percabangan di dalam loop).

- Tantangan: Batasi total iterasi valid maksimal 10 angka. Jika sudah mencapai 10 angka valid, loop harus berhenti otomatis meskipun belum ada angka 999.

```python
jumlah = 0
batasAngka = 10
validasi = 0

# Looping
while validasi < batasAngka:

    # user input
    user = int(input("Enter Your name : "))

    # Conditional Formatting
    if user <= 0:
        continue

    elif user == 999:
        break

    elif user % 5 == 0:
        continue

    else:
        jumlah = jumlah + user
        validasi += 1
        print(jumlah)
```

# 3. Pencari Bilangan Prima Berkinerja Tinggi dengan Optimasi (Hard)

Deskripsi : Buatlah program berbasis while loop untuk mencari dan mencetak semua bilangan prima dari angka 2 hingga N (di mana N diinputkan oleh pengguna, misal N = 100).

- Gunakan while loop untuk mengecek apakah sebuah angka adalah bilangan prima.
- Di dalam pengecekan faktor pembagi, jika ditemukan faktor selain 1 dan angka itu sendiri, segera hentikan pengecekan angka tersebut menggunakan break untuk menghemat sumber daya CPU (early exit).
- Untuk optimasi, gunakan continue agar loop mengabaikan pengecekan pada angka genap di atas 2 (karena sudah pasti bukan prima).

```python
# Bilangan prima

angka = 2
inputNilai = int(input("Masukkan angka : "))

# Looping
while angka <= inputNilai:

    # Conditional Formatting
    if angka < 2 and angka % 2 == 0:
        angka += 1
        continue

    # Pembagi dan Result
    pembagi = 2
    prima = True

    while pembagi < angka:
        if angka % pembagi == 0:
            prima = False
            break

        pembagi += 1

    if prima:
        print(angka)

    angka += 1
```
