```python
# Constructor 

class mobil:
    def __init__(self, nama, tahun, merek):
        self.nama = nama,
        self.tahun = tahun,
        self.merek = merek,

y = mobil('Bendu', 1987, 'BMW audi 445')
print(y.nama)
print(y.tahun)
print(y.merek)
print(y)
```

```python
## Class And Object

# How To Create a Class

class Race:
    r = 5

class mobil:
    name = "BMW Audi 6000"

class dragRace:
    hg = 4.5
```

```python
# How To Create Object

drag = Race()
print(drag.r)

klasik = mobil()
print(klasik.name)

ukuranRace = dragRace()
print(ukuranRace.hg)
```
