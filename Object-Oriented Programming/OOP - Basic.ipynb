```python
# Constructor 

class mobil:
    def __init__(self, nama, tahun, merek):
        self.nama = nama,
        self.tahun = tahun,
        self.merek = merek,

y = mobil('Bendu', 1987, 'BMW audi 445')
print(y.nama)
print(y.tahun)
print(y.merek)
print(y)
```
